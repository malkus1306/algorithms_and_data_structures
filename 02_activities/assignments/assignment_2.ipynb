
-   Copy the solution your partner wrote. 


```python
# Your answer here
from collections import deque
from typing import List

# Define the TreeNode class
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def bt_path(root: TreeNode) -> List[List[int]]:
    if not root:
        return []

    result = []
    queue = deque([(root, [root.val])])  # Store node and path so far

    while queue:
        node, path = queue.popleft()

        # If it's a leaf node, add the path to result
        if not node.left and not node.right:
            result.append(path)

        # Add left child to queue if it exists
        if node.left:
            queue.append((node.left, path + [node.left.val]))

        # Add right child to queue if it exists
        if node.right:
            queue.append((node.right, path + [node.right.val]))

    return result

# Example Usage
# Construct the binary tree: [1, 2, 2, 3, 5, 6, 7]
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(2)
root.left.left = TreeNode(3)
root.left.right = TreeNode(5)
root.right.left = TreeNode(6)
root.right.right = TreeNode(7)

# Get all paths from root to leaf nodes
output = bt_path(root)
print(output)
```


-   Explain why their solution works in your own words.


```python
# Your answer here
# The partner's solution is based on the class TreeNode and specifically focuses on the method bt_path. 
# This method uses Breadth First Search (BFS) approach to traverse the tree.
# BFS helps to identify all the paths from root to the leaves.
# More specifically, we are exploring every node once adding it the queue until the leaf node is approached. 
# When the leaf node is approached the specific path from the queue is added to the result. 
```

```python
# Your answer here
# In case we have n nodes in the tree:
# Time complexity is O(n) as every node is explored 1 time during the BFS
# Space complexity is O(n) as in the worst case scenario there could be up to n/2 nodes at the last level to store
```
