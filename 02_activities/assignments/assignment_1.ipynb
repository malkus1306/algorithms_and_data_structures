```python
# Your answer here
def missing_nums(nums: List):
    # Sort the list
    sorted = merge_sort(nums)    
    result = []
    qty = len(sorted) - 1
    if qty == -1:
        return -2 # alternative result in case of wrong input

    if qty > 0:
        # Compare number of elements with the last (maximum) value in the provided list
        if qty > sorted[-1]:
            result = list(range(sorted[-1] + 1, qty + 1)) + result
            if sorted[-1] - sorted[-2] > 1:
                result = list(range(sorted[-2] + 1, sorted[-1])) + result
        elif qty <= sorted[-1]:
            result = list(range(sorted[-2] + 1, sorted[-1])) + result
        previous = sorted[-1]
        sorted.pop()

        # In case of any omitted integer add it to the resulting list
        while len(sorted) > 1:
            if sorted[-1] - sorted[-2] > 1:
                result = list(range(sorted[-2] + 1, sorted[-1])) + result
            previous = sorted[-1]
            sorted.pop()
            
        if sorted[0] > 0:
            result = list(range(0, sorted[0])) + result
    
    else:
        if sorted[0] == 0:
            return -1
        else:
            return list(range(0, sorted[0]))

    if len(result) == 0:
        return -1
    else:
        return result

def merge_sort(nums: List):
    if len(nums) <= 1:
        return nums

    middle = len(nums) // 2
    left = merge_sort(nums[:middle])
    right = merge_sort(nums[middle:])

    return merge(left, right)

def merge(left, right):
    sorted = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted.append(left[i])
            i += 1
        else:
            sorted.append(right[j])
            j += 1

    sorted.extend(left[i:])
    sorted.extend(right[j:])

    return sorted
```
